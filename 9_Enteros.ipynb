# Números Enteros $\mathbb{Z}$: Construcción Formal e Intuición

Los números enteros ($\mathbb{Z}$) surgen como una ampliación necesaria de los números naturales ($\mathbb{N}$) para poder expresar deudas, carencias o "cosas que no tenemos".

En $\mathbb{N}$, la expresión algebraica:
$$b + x = a$$
no tiene solución cuando $a < b$. Es decir, no existe un $x \in \mathbb{N}$ que satisfaga dicha ecuación, ya que en los números naturales no existen los números negativos (sus elementos opuestos).

Cuando $b \le a$, el valor de $x \in \mathbb{N}$ es **único**. Sin embargo, si representamos este escenario mediante una tupla $(a, b)$, nos damos cuenta de que existen **infinitos pares** $a, b \in \mathbb{N}$ que corresponden a un mismo valor de $x$.

### Intuición: "Tengo" y "Debo"
Podemos interpretar la tupla $(a, b)$ como **(tengo, debo)**. Esta expresa nuestro balance entre propiedades y deudas.

Por ejemplo, tener 5 manzanas y deber 2 nos deja un balance neto de 3. Esto es exactamente lo mismo que tener 100 manzanas y deber 97; el balance neto ($x$) siempre es el mismo en ambos escenarios.
Es decir, las tuplas $(5, 2)$ y $(100, 97)$ representan intrínsecamente el mismo número natural $3 \in \mathbb{N}$.

## Relación de Equivalencia $\mathcal{E}$ sobre $\mathbb{N} \times \mathbb{N}$

Dado que múltiples tuplas $(a, b)$ representan el mismo valor $x$ en la ecuación $b + x = a$, podemos establecer una **relación de equivalencia**.

Si representamos lo que "tengo" en el eje $x$ y lo que "debo" en el eje $y$, obtenemos que $(a, b) \in \mathbb{N}^2$. Cada tupla se convierte en un punto en el plano cartesiano $\mathbb{N} \times \mathbb{N}$.

Sean $(a, b)$ y $(a', b')$ dos tuplas (tengo, debo) cualesquiera en $\mathbb{N}^2$. El siguiente predicado define su relación de equivalencia formal:

$$(a, b) \mathcal{E} (a', b') \iff a + b' = a' + b$$

Esta expresión define una relación de equivalencia sobre el conjunto $\mathbb{N} \times \mathbb{N}$.

## El Conjunto Cociente: La Definición de $\mathbb{Z}$

Basándonos en la relación anterior, el **conjunto cociente** $\mathbb{N}^2 / \mathcal{E}$ es, por definición, el conjunto de los números enteros:

$$\frac{\mathbb{N} \times \mathbb{N}}{\mathcal{E}} = \mathbb{Z}$$

Por lo tanto, una **clase de equivalencia** $[(a, b)] \in \mathbb{N}^2 / \mathcal{E}$ es, por definición formal, un número entero:

$$[(a, b)] \in \mathbb{Z}$$

### Intuición
Un número entero es en realidad una **clase** o partición de un conjunto cociente generada por la relación de equivalencia. Una clase es en sí misma un conjunto de elementos equivalentes entre sí (todas las tuplas que dan el mismo balance neto). Por lo tanto, un número entero se puede entender estructuralmente como un conjunto de tuplas.

## Representantes Canónicos ($\alpha$)

Dentro del conjunto infinito de tuplas que forman un número entero $[(a, b)]$, existe un elemento privilegiado que representa a todo el conjunto. Como vimos en la gráfica, cada diagonal puede ser identificada de forma única por un punto base; este representante se llama **representante canónico** $\alpha$.

Se define como:
$$\alpha = (a, b) \in [(a, b)] \mid a = 0 \lor b = 0$$

### Intuición Geométrica
Partiendo de cualquier punto en $\mathbb{N}^2$, puedo "deslizarme" por su diagonal correspondiente $[(a, b)]$ hasta interceptar el eje $x$ o el eje $y$. El punto exacto donde la diagonal corta uno de estos ejes es $\alpha$, nuestro representante canónico.

El hecho de que $\alpha$ deba tener la forma $(m, 0)$, $(0, n)$ o $(0,0)$ tiene implicaciones algebraicas fundamentales para clasificar los números enteros.

## Clasificación: Enteros Positivos y Negativos

Dependiendo del eje en el que recaiga el representante canónico $\alpha$, podemos clasificar los enteros en positivos o negativos.

### 1. Enteros Positivos: $\alpha = (m, 0)$
Los representantes que intersectan sobre el **eje $x$** son los enteros positivos.



positivos.png


> **Nota descriptiva:** Gráfica cartesiana enfocada en diagonales que nacen a partir del eje X, representando los valores donde "debo" es 0 inicialmente.

Para que una tupla pertenezca a esta clase de equivalencia:
$$(a, b) \mathcal{E} (m, 0) \iff a + 0 = m + b \iff a = m + b \implies b \le a$$

* **Intuición:** Si $a$ = "tengo" y $b$ = "debo", la condición $b \le a$ significa que **debo menos o igual de lo que tengo** (o tengo más de lo que debo). El resultado es un "entero positivo".

### 2. Enteros Negativos: $\alpha = (0, n)$
Los representantes que intersectan sobre el **eje $y$** son los enteros negativos.


negativos.png


> **Nota descriptiva:** Gráfica cartesiana enfocada en diagonales que nacen a partir del eje Y, representando los valores donde "tengo" es 0 inicialmente.

Para que una tupla pertenezca a esta clase de equivalencia:
$$(a, b) \mathcal{E} (0, n) \iff a + n = 0 + b \iff b = n + a \implies a \le b$$

* **Intuición:** Siguiendo la misma analogía, la condición $a \le b$ significa que **tengo menos o igual a lo que debo** (debo más de lo que tengo). El resultado es un "entero negativo".

```julia
# Definirmos el producto: A×B={(x,y)|x∈A,y∈B}
function ×(A, B)
     return Set((x, y) for x ∈ A for y ∈ B)
end

# Convertir las relaciones a diccionarios:
## k=representante canonico, v= Clase de equivalencia
function canonical_repre(
    𝓔::Set{Tuple{Tuple{T,T},Tuple{T,T}}},
    alphas::Set{Tuple{T, T}},
    ) where {T}
    canonical_classes = Dict{
        Tuple{T,T}, Set{Tuple{T,T}}
        }() # Vacio
    # Representante canonico α
    # α=(a,b) | a=0 ∨ b=0, (a,b)𝓔(a´,b´)
    for α ∈ alphas
        # [α]={(a´,b´) | α𝓔(a´,b´)}
        for ((a,b),(a´,b´)) ∈ 𝓔
            if α == (a,b)
                push!(
                    get!(canonical_classes, α, Set{T}()),
                    (a´,b´)
                    ) # Insert new k:v
            end
        end
    end
    return canonical_repre
end
```

```julia
ℕ = 0:10
var"ℕ²" = ℕ × ℕ
# (a,b)𝓔(a´,b´) ⟺ a+b´=a´+b
predicado(
    tup1::Tuple{Int, Int}, tup2::Tuple{Int, Int}
    ) = tup1[1] + tup2[2] == tup2[1] + tup1[2]

𝓔 = Set([
    ((a,b),(a´,b´))
    for (a,b) ∈ var"ℕ²", (a´,b´) ∈ var"ℕ²"
    if predicado((a,b), (a´,b´))
    ])
# α=(a,b) | a=0 ∨ b=0, (a,b)𝓔(a´,b´)
alphas = Set([tup for tup ∈ var"ℕ²" if tup[1]==0 || tup[2]==0])
# [α] ∈ ℕ×ℕ/𝓔
# ℤ = ℕ×ℕ/𝓔
var"[α]" = canonical_classes(𝓔, alphas)
```

```julia
using Plots

# 1. Preparar el plot base
plt = plot(title="Clases Canonicas", legend = :outerright)

# 2. Dibujar las líneas y recolectar información para los ejes
x_labels_pos = Int64[]
x_labels_text = String[]
y_labels_pos = Int64[]
y_labels_text = String[]

for (key, puntos) in var"[α]"
    # Ordenar y extraer coordenadas
    sorted_pts = sort(collect(puntos))
    xs = [p[1] for p in sorted_pts]
    ys = [p[2] for p in sorted_pts]

    # Graficar la línea de la clase
    if key == (0,0)
        plot!(plt, xs, ys, label="Clase [$key]", marker=:circle, color=:black)
    else
        plot!(plt, xs, ys, label="Clase [$key]", marker=:circle)
    end

    # Identificar si la key es una marca para el eje X o el eje Y
    kx, ky = key
    if ky == 0
        push!(x_labels_pos, kx)
        push!(x_labels_text, "($(kx),0)")
    end
    if kx == 0
        push!(y_labels_pos, ky)
        push!(y_labels_text, "(0,$(ky))")
    end
end
# 4. Aplicar las etiquetas personalizadas a los ejes
# Esto cambia los números simples por la notación de la 'key'
xticks!(plt, x_labels_pos, x_labels_text)
yticks!(plt, y_labels_pos, y_labels_text)

display(plt)
```

# Operaciones en $\mathbb{Z}$: Suma y Producto

Las operaciones de suma ($+$) y producto ($\cdot$) son heredadas por los números naturales ($\mathbb{N}$), ya que, por definición, un entero es una clase de equivalencia formada por pares de números naturales ($\mathbb{N} \times \mathbb{N}$). La única novedad fundamental respecto a los naturales ($\mathbb{N}$) es la aparición de los elementos **opuestos** (números negativos).

Sean $\alpha, \beta \in \mathbb{Z}$. Estos son clases del conjunto cociente $\frac{\mathbb{N} \times \mathbb{N}}{\mathcal{E}}$ que podemos representar mediante tuplas genéricas:
$$\alpha = [(a,b)] \quad \text{y} \quad \beta = [(c,d)]$$

## Intuición: El Modelo de "Contabilidad Doble"

Podemos visualizar estas operaciones mediante un sistema de contabilidad doble: por un lado se registran **ganancias** (lo que "tengo") y por otro **pérdidas/deudas** (lo que "debo").
* $\alpha = [(a,b)]$ representa una transacción donde gano $a$ y debo $b$.
* $\beta = [(c,d)]$ representa una transacción donde gano $c$ y debo $d$.

### La Suma ($+$)
Sumar en nuestra contabilidad es simplemente agrupar todas las ganancias por un lado y todas las deudas por el otro:
$$\alpha + \beta = [(a,b)] + [(c,d)] = [(\underbrace{a+c}_{\text{ganar}}, \underbrace{b+d}_{\text{deber}})]$$
El balance final consta de unas ganancias totales de $a+c$ y unas deudas totales de $b+d$.

### El Producto ($\cdot$)
La definición del producto es un poco más compleja, pero se deduce lógicamente del mismo modelo. Imagina que el producto significa que "tengo" $c$ y "debo" $d$ "cajas", donde cada caja contiene la contabilidad interna de $\alpha$ (gano $a$, debo $b$).

$$\alpha \cdot \beta = [(a,b)] \cdot [(c,d)] = [(\underbrace{a\cdot c + b\cdot d}_{\text{ganancia neta}}, \underbrace{b\cdot c + a\cdot d}_{\text{deuda neta}})]$$

**Desglose intuitivo:**
* **Ganancias netas ($a\cdot c + b\cdot d$):**
  * $a\cdot c$: Si **tengo** $c$ cajas y en cada una hay $a$ ganancias, obtengo $a\cdot c$ de ganancia.
  * $b\cdot d$: Si **pierdo/debo** $d$ cajas y en cada una hay $b$ deudas, al quitarme deudas, genero una ganancia equivalente a $b\cdot d$. *"perder deuda es una ganancia"*
* **Pérdida/Deuda neta ($b\cdot c + a\cdot d$):**
  * $b\cdot c$: Si **tengo** $c$ cajas y en cada una hay $b$ deudas, acumulo una deuda total de $b\cdot c$.
  * $a\cdot d$: Si **pierdo** $d$ cajas y en cada una hay $a$ ganancias, perder mis bienes/ganancias se traduce en una deuda o pérdida neta de $a\cdot d$.

Esta intuición contable no es más que la aplicación natural de la **propiedad distributiva** sobre los componentes de la tupla.

## Independencia del Representante $\alpha = [(a,b)]$

Es fundamental destacar que la suma y el producto en $\mathbb{Z}$ no se definen sobre el conjunto completo $[(a,b)]$, sino sobre **uno de sus elementos** (una tupla específica). Por lo tanto, para que la matemática sea sólida, las operaciones deben arrojar el mismo resultado independientemente del elemento elegido para representar la clase.

**Ejemplo Intuitivo:** Imagina a dos personas, A y B, que manejan contabilidades diferentes pero cuyo balance neto a fin de mes es exactamente el mismo.

| Persona A | (Tengo $a$, Debo $b$) | | Persona B | (Tengo $a$, Debo $b$) |
| :--- | :--- | :--- | :--- | :--- |
| Mes 1 | $(100, 30)$ | | Mes 1 | $(120, 50)$ |
| Mes 2 | $(40, 90)$ | | Mes 2 | $(10, 60)$ |
| **Suma $(+)$** | $[(100+40, 30+90)]$ | | **Suma $(+)$** | $[(120+10, 50+60)]$ |
| **Total** | $[(140, 120)]$ | | **Total** | $[(130, 110)]$ |
| **Neto ($x$)** | $140 = 120 + x \implies x = 20$ | | **Neto ($x$)** | $130 = 110 + x \implies x = 20$ |

Aunque las tuplas totales son distintas ($(140, 120)$ frente a $(130, 110)$), ambas pertenecen a la misma clase de equivalencia, representada por su balance canónico $[(20, 0)]$:
$$(140, 120) \mathcal{E} (130, 110) \mathcal{E} (20, 0)$$

Formalmente, si $(mes_{1A} + mes_{2A}) \mathcal{E} (mes_{1B} + mes_{2B})$, entonces:
$$((a,b) + (c,d)) \mathcal{E} ((a',b') + (c',d'))$$
De forma similar ocurre con el producto:
$$((a,b) \cdot (c,d)) \mathcal{E} ((a',b') \cdot (c',d'))$$

## Suma de Positivos, Negativos y Ley de Signos

Como vimos previamente, los **enteros positivos ($\mathbb{Z}_+$)** se representan con tuplas de la forma $[(m, 0)]$, mientras que los **enteros negativos ($\mathbb{Z}_-$)** con tuplas $[(0, n)]$.

**1. Suma de elementos del mismo signo:**
* Sumar positivos (ganancias) siempre da ganancias: $[(m, 0)] + [(m', 0)] = [(m+m', 0)]$
* Sumar negativos (deudas) siempre da deudas: $[(0, n)] + [(0, n')] = [(0, n+n')]$

**2. Suma de signos opuestos:**
¿Qué ocurre al sumar $[(m, 0)] + [(0, n)]$? El resultado depende de cuál de los dos valores sea mayor (si la deuda o la ganancia):
$$[(m, 0)] + [(0, n)] = [(m, n)]$$
* Si $n < m \implies [(m, n)] \in \mathbb{Z}_+$ (positivo)
* Si $m < n \implies [(m, n)] \in \mathbb{Z}_-$ (negativo)

**3. Ley de Signos en el Producto:**
Al aplicar nuestra definición de producto, la regla de los signos emerge de forma natural y evidente:
* $(+) \cdot (+): \quad [(m,0)] \cdot [(m',0)] = [(m\cdot m' + 0, 0 + 0)] = [(mm', 0)] \implies (+)$
* $(-) \cdot (-): \quad [(0,n)] \cdot [(0,n')] = [(0 + n\cdot n', 0 + 0)] = [(nn', 0)] \implies (+)$
* $(+) \cdot (-): \quad [(m,0)] \cdot [(0,n)] = [(m\cdot 0 + 0\cdot n, 0 + m\cdot n)] = [(0, mn)] \implies (-)$

## Estructura Algebraica: El Grupo Abelian $(\mathbb{Z}, +)$

Respecto a la operación interna de suma $+$, el conjunto $\mathbb{Z}$ forma un **grupo conmutativo** (o abeliano) ya que cumple las siguientes propiedades para todo $\alpha, \beta, \gamma \in \mathbb{Z}$:

**1. Asociatividad:**
$$(\alpha + \beta) + \gamma = \alpha + (\beta + \gamma)$$
*Demostración:*
$$([(a,b)] + [(c,d)]) + [(e,f)] = [(a+c, b+d)] + [(e,f)] = [(a+c+e, b+d+f)]$$
$$= [(a,b)] + [(c+e, d+f)] = \alpha + (\beta + \gamma)$$

**2. Elemento Neutro:**
Existe un elemento $[(0,0)]$ tal que:
$$[(m,n)] + [(0,0)] = [(m+0, n+0)] = [(m,n)]$$

**3. Elemento Simétrico (Opuesto):**
Para cada entero existe un opuesto denotado como $-\alpha$.
$$[(m, 0)] + [(0, m)] = [(m+0, 0+m)] = [(m, m)]$$
Dado que la tupla $(m,m)$ significa "ganar $m$ y deber $m$", el balance neto es 0. Por tanto, $[(m,m)] \mathcal{E} [(0,0)]$.
En notación algebraica: $-[(m,0)] = [(0,m)]$.

**4. Conmutatividad:**
$$\alpha + \beta = \beta + \alpha$$
*Demostración:*
$$[(a,b)] + [(c,d)] = [(a+c, b+d)] = [(c+a, d+b)] = [(c,d)] + [(a,b)] = \beta + \alpha$$

## Estructura Algebraica: El Anillo $(\mathbb{Z}, +, \cdot)$

Al incorporar el producto, el conjunto $\mathbb{Z}$ adquiere la estructura de un **anillo**, cumpliendo con las siguientes propiedades adicionales para $\alpha, \beta, \gamma \in \mathbb{Z}$:

**1. Asociatividad del Producto:**
$$(\alpha \cdot \beta) \cdot \gamma = \alpha \cdot (\beta \cdot \gamma)$$
*Demostración:*
$$([(a,b)] \cdot [(c,d)]) \cdot [(e,f)] = [(ac+bd, ad+bc)] \cdot [(e,f)]$$
$$= [((ac+bd)e + (ad+bc)f, (ac+bd)f + (ad+bc)e)]$$
$$= [(ace+bde+adf+bcf, acf+bdf+ade+bce)]$$
Reagrupando términos, esto es idéntico a calcular $[(a,b)] \cdot ([(c,d)] \cdot [(e,f)])$.

**2. Distributividad del Producto respecto a la Suma:**
$$\alpha \cdot (\beta + \gamma) = \alpha \cdot \beta + \alpha \cdot \gamma$$
*Demostración:*
$$[(a,b)] \cdot ([(c,d)] + [(e,f)]) = [(a,b)] \cdot [(c+e, d+f)]$$
$$= [(a(c+e) + b(d+f), a(d+f) + b(c+e))]$$
$$= [(ac+ae+bd+bf, ad+af+bc+be)]$$
Y por el otro lado:
$$[(ac+bd, bc+ad)] + [(ae+bf, be+af)] = [(ac+bd+ae+bf, bc+ad+be+af)]$$
Ambos resultados son idénticos.

**3. Conmutatividad del Producto:**
$$\alpha \cdot \beta = \beta \cdot \alpha$$
*Demostración:*
$$[(a,b)] \cdot [(c,d)] = [(ac+bd, bc+ad)] = [(ca+db, da+cb)] = [(c,d)] \cdot [(a,b)]$$

**4. Elemento Neutro Multiplicativo:**
Para que $\mathbb{Z}$ sea un anillo unitario, el elemento neutro del producto es **$[(1,0)]$**.
$$[(m,n)] \cdot [(1,0)] = [(m\cdot 1 + n\cdot 0, n\cdot 1 + m\cdot 0)] = [(m,n)]$$

Por todo lo anterior, **$(\mathbb{Z}, +, \cdot)$ es un Anillo Conmutativo Unitario**.

## Dominio de Integridad (Anillo Íntegro)

Por último, $(\mathbb{Z}, +, \cdot)$ es un **anillo íntegro** (o dominio de integridad). Esto significa que no existen divisores de cero; es decir, no existen dos números enteros distintos de cero que al multiplicarse den cero.

Formalmente, es falso que:
$$\alpha \neq 0 \land \beta \neq 0 \mid \alpha \cdot \beta = 0 \iff \text{Falso}$$

En otras palabras, para que la multiplicación sea cero, obligatoriamente alguno de los operandos debe ser cero:
$$\alpha \cdot \beta = 0 \implies \alpha = 0 \lor \beta = 0$$

# Relación de Orden en $\mathbb{Z}$

Para determinar si un número entero es positivo o negativo (es decir, si pertenece a $\mathbb{Z}_+$ o a $\mathbb{Z}_-$), recurrimos a la definición de orden heredada de los números naturales ($\mathbb{N}$), teniendo en cuenta que una clase $[(a,b)]$ está formada por $a, b \in \mathbb{N}$.

### 1. Enteros Positivos ($\mathbb{Z}_+$)
Si $b \le a$, entonces la clase pertenece a los enteros positivos:
$$b \le a \implies [(a,b)] \in \mathbb{Z}_+$$

Por definición de orden en $\mathbb{N}$, que $b \le a$ significa que existe un $x \in \mathbb{N}$ tal que $b + x = a$.
* **Interpretación:** Las ganancias ($a$) son mayores que las deudas ($b$). Se necesita un total de $x$ para igualarlas. Si calculamos el neto, nos quedan $x$ ganancias y $0$ deudas.
* Por tanto, el **representante canónico** de todos los números positivos tiene la forma $[(m,0)]$.

Formalmente, el conjunto de los enteros positivos es:
$$\mathbb{Z}_+ = \{[(m,0)] \in \mathbb{Z} \mid m \in \mathbb{N}\}$$

### 2. Enteros Negativos ($\mathbb{Z}_-$)
De forma análoga, si $a \le b$, la clase pertenece a los enteros negativos:
$$a \le b \implies [(a,b)] \in \mathbb{Z}_-$$

Esto implica que $\exists x \in \mathbb{N}$ tal que $a + x = b$.
* **Interpretación:** Las deudas ($b$) superan a las ganancias ($a$). El neto en este caso indica $0$ ganancias y $x$ deudas.
* El **representante canónico** de los números negativos es $[(0,n)]$.

Formalmente, el conjunto de los enteros negativos es:
$$\mathbb{Z}_- = \{[(0,n)] \in \mathbb{Z} \mid n \in \mathbb{N}\}$$

## Comparación entre Clases: ¿Qué elemento precede a otro?

Comparando cada elemento de la tupla (gano, debo), logramos organizar a $\mathbb{Z}$ en los dos grandes grupos ($\mathbb{Z}_+$ y $\mathbb{Z}_-$). Pero dentro de un mismo grupo, ¿cómo sabemos qué elemento es mayor?

La **regla general** de orden para dos enteros cualesquiera $\alpha = [(a,b)]$ y $\beta = [(c,d)] \in \mathbb{Z}$ es:

$$\alpha \le \beta \iff \beta - \alpha \in \mathbb{Z}_+$$

**Desarrollo algebraico:**
Recordemos que restar $\alpha$ es sumar su opuesto $-\alpha = [(b,a)]$:
$$\beta - \alpha = [(c,d)] - [(a,b)] = [(c,d)] + [(b,a)] = [(c+b, d+a)]$$

Por lo tanto:
$$\alpha \le \beta \iff [(c+b, d+a)] \in \mathbb{Z}_+$$

Para que esto se cumpla, las "ganancias" resultantes ($c+b$) deben ser mayores o iguales que las "deudas" resultantes ($d+a$):
* Si $d+a \le c+b \implies [(c+b, d+a)] \in \mathbb{Z}_+ \iff \alpha \le \beta$
* Si $c+b \le d+a \implies [(c+b, d+a)] \notin \mathbb{Z}_+ \iff [(c+b, d+a)] \in \mathbb{Z}_- \iff \beta \le \alpha$

> **Intuición:** Si tengo dos acciones/transacciones financieras $(a,b)$ y $(c,d)$, ¿cómo sé cuál me da más ganancia neta? Esto ocurre si una tiene mucha ganancia (e.g., $c$) y la otra mucha deuda ($b$). Por tanto, si la *ganancia de una más la deuda de la otra* ($c+b$) es mayor que la *deuda de la primera más la ganancia de la segunda* ($d+a$), es seguro que $(c,d) \ge (a,b)$.

## Propiedades de la Relación de Orden en $\mathbb{Z}$

La relación definida como $\beta - \alpha \in \mathbb{Z}_+$ cumple con las tres propiedades fundamentales de una relación de orden:

1.  **Reflexiva:** $$\alpha - \alpha \in \mathbb{Z}_+$$
    *Demostración:* $[(a,b)] - [(a,b)] = [(a,b)] + [(b,a)] = [(a+b, b+a)] \mathcal{E} [(0,0)]$. Como $[(0,0)] \in \mathbb{Z}_+$ (el cero se asume no-negativo), se cumple.
2.  **Antisimétrica:** $$\alpha \le \beta \land \beta \le \alpha \implies \beta - \alpha \in \mathbb{Z}_+ \cap \mathbb{Z}_- = \{[(0,0)]\}$$
Por tanto, el neto es cero, lo que implica que $\alpha = \beta$.
3.  **Transitiva:** $$\alpha \le \beta \land \beta \le \gamma \implies \alpha \le \gamma \iff \gamma - \alpha \in \mathbb{Z}_+$$

Además, esta relación dota a $\mathbb{Z}$ de dos características esenciales:
* **Orden Total:** Cualquier par de elementos es comparable, ya que $\mathbb{Z}_+ \cup \mathbb{Z}_- = \mathbb{Z}$.
* **Compatibilidad con la Suma ($+$):** Si $\alpha \le \beta$, entonces $\gamma + \alpha \le \gamma + \beta$.
    *Demostración:* $\gamma + \beta - (\gamma + \alpha) = \gamma + \beta - \gamma - \alpha = \beta - \alpha \in \mathbb{Z}_+ \iff \alpha \le \beta$.

## Valor Absoluto en $\mathbb{Z}$

Dado que $(\mathbb{Z}, +, \cdot, \le)$ tiene un orden total, se le considera un anillo totalmente ordenado. Esto permite definir formalmente la función valor absoluto para cualquier $\alpha \in \mathbb{Z}$:

$$
|\alpha| =
\begin{cases}
\alpha & \text{si } [(0,0)] \le \alpha \\
-\alpha & \text{si } \alpha < [(0,0)]
\end{cases}
$$

Es decir:
* Para los positivos: $|[(m,0)]| = [(m,0)]$
* Para los negativos: $|[(0,n)]| = -[(0,n)] = [(n,0)]$

### Propiedades e Intuición del Valor Absoluto

Sean $\alpha, \beta \in \mathbb{Z}$, se cumplen las siguientes propiedades:

**1. Multiplicatividad:** $|\alpha \cdot \beta| = |\alpha| \cdot |\beta|$
> *Intuición:* Es lo mismo dejar que ocurra toda la operación multiplicativa (ganar o traspasar deudas/ganancias) y después mirar el balance neto ignorando su dirección (signo), que mirar el impacto del "paquete financiero" $|\alpha|$ individualmente, evaluar el volumen neto de transacciones $|\beta|$, y multiplicarlos.

**2. Desigualdad Triangular:** $|\alpha + \beta| \le |\alpha| + |\beta|$
> *Intuición:* El valor $|\alpha + \beta|$ calcula el neto de la operación conjunta sin importar si fue pérdida o ganancia. Como estamos sumando *antes* de aplicar el valor absoluto, ya estamos permitiendo que las ganancias cancelen a las pérdidas internamente. En cambio, en $|\alpha| + |\beta|$, estamos calculando el neto individual de cada transacción asumiéndolas todas como ganancias (ignorando su naturaleza real), sumándolas sin permitir que se cancelen entre sí. Por ende, la suma separada siempre será igual o mayor.

## Identificación de $\mathbb{N}$ con $\mathbb{Z}_+$ (Isomorfismo)

La identificación natural que hacemos de los números naturales con los enteros positivos ocurre mediante un **homomorfismo** de anillos ordenados:
$$f: (\mathbb{N}, +, \cdot, \le) \longrightarrow (\mathbb{Z}, +, \cdot, \le)$$

Para las diagonales de la parte superior izquierda (clases $[(0,n)] \in \mathbb{Z}_-$), simplemente se reflejan en el eje X negativo en orden inverso.

Dicho homomorfismo se convierte formalmente en un **Isomorfismo** (una biyección perfecta que preserva la estructura) si restringimos el codominio exclusivamente a $\mathbb{Z}_+$:
$$f: (\mathbb{N}, +, \cdot, \le) \longrightarrow (\mathbb{Z}_+, +, \cdot, \le)$$

De esta forma es evidente que para todo $a, b \in \mathbb{N}$:
1.  $f(a+b) = f(a) + f(b)$
2.  $f(a \cdot b) = f(a) \cdot f(b)$
3.  $a \le b \implies f(a) \le f(b)$

De forma que los enteros negativos son, por construcción, simplemente los opuestos de las proyecciones de $\mathbb{N}$ en $\mathbb{Z}$:
* $m \in \mathbb{N} \implies f(m) = [(m,0)]$
* $n \in \mathbb{N} \implies -f(n) = -[(n,0)] = [(0,n)]$

### El "Abuso de Notación"
Debido a este nivel de abstracción y homomorfismo, solemos expresar la clase $[(m,0)]$ simplemente como $m$ y a $[(0,n)]$ como $-n$. Es vital entender que matemáticamente esto es un **abuso de notación**. Ocurre que $f(\mathbb{N}) \in \mathbb{Z}_+ \subset \mathbb{Z}$, pero hablando con rigor lógico-conjuntista, **no es verdad que $\mathbb{N} \subset \mathbb{Z}$**. Son conjuntos construidos con naturalezas distintas (números base vs. clases de equivalencia de pares de números base).

## Buen Orden en $\mathbb{Z}$

Debido al isomorfismo mencionado, $\mathbb{Z}_+$ hereda todas las propiedades estructurales de $\mathbb{N}$, entre estas, la noción de buen orden:
1.  Todo subconjunto $A \subset \mathbb{Z}$ tal que $A \neq \emptyset$ y que esté **acotado superiormente**, tiene un elemento máximo absoluto: $\max(A) \in A$.
2.  Si el subconjunto $A$ está **acotado inferiormente**, tiene un elemento mínimo: $\min(A) \in A$.

Esto aplica por extensión a $\mathbb{Z}_-$, ya que se comportan estructuralmente igual que $\mathbb{N}$ pero en un orden invertido a través del espejo del cero (por lo que el límite del máximo pasa a ser el del mínimo y viceversa).

## Propiedad Arquimediana de $\mathbb{Z}$

Sean $\alpha, \beta \in \mathbb{Z}$. Esta propiedad garantiza que no importa qué tan lejos esté un número $\alpha$ de un objetivo $\beta$ dentro de nuestra relación de orden; siempre es posible alcanzar o sobrepasar a $\beta$ acumulando un número finito $n$ de pasos de tamaño $\alpha$.

Por debajo, se ven "saltos" sucesivos de un tamaño fijo $\alpha$. Tras $n$ saltos multiplicativos ($n \cdot \alpha$), la trayectoria rebasa la posición original de $\beta$.

En otras palabras, con finitos pasos es posible alcanzar cualquier número en $\mathbb{Z}$. Algebraicamente se expresa como:
$$\forall \alpha, \beta \in \mathbb{Z}, \text{ con } \alpha > 0 \implies \exists n \in \mathbb{N} \text{ tal que } n \cdot \alpha \ge \beta$$

**Demostración lógica (Esbozo):**
Si construimos el conjunto de todos los pasos posibles $k$ de tamaño $\alpha$ que aún no alcanzan o apenas igualan a $\beta$, tendríamos:
$$A = \{k \cdot \alpha \mid k \cdot \alpha \le \beta, k \in \mathbb{N}\}$$

Por el axioma del buen orden, al estar $A$ acotado superiormente por $\beta$, garantizamos que existe un máximo en dicho conjunto: $\max(A) = m \cdot \alpha$.
* Esto garantiza por definición que $m \cdot \alpha \le \beta$.
* Por tanto, su inmediato sucesor $(m+1) \cdot \alpha$ obligatoriamente queda excluido del conjunto $A$: $(m+1) \cdot \alpha \notin A$.
* Si no pertenece al conjunto, es estrictamente mayor al límite: $(m+1) \cdot \alpha > \beta$.

De este modo comprobamos que siempre existirá un múltiplo finito que cumpla la propiedad de rebasar cualquier cota impuesta.

# División Entera en $\mathbb{Z}$

La **Propiedad Arquimediana** garantizaba que, con un número suficiente de pasos $n$, siempre podemos alcanzar o superar un objetivo $\beta$: $\exists n \in \mathbb{N} \mid n \cdot \alpha \ge \beta$.

Pero, ¿qué pasa si queremos acercarnos lo máximo posible a un número sin sobrepasarlo? Este es el caso fundamental de la **división entera**. Dados $a, b \in \mathbb{Z}$ con $b \neq 0$:

$$q \cdot b \le a \iff a = q \cdot b + r \land 0 \le r < |b|$$

* $q$ es llamado **cociente**, y significa el "número de veces" o pasos de tamaño $b$ que podemos dar.
* $r$ es el **resto**, que representa lo que faltó para llegar exactamente a $a$.

### Propiedades del Resto e Intuición
* **¿Por qué $r \neq |b|$?** Si $r = |b|$, significaría que exactamente nos faltó un paso más de tamaño $b$, por lo que $q$ podría haber sido mayor.
* **¿Por qué $r < |b|$?** Si $r > |b|$, $r$ no sería el resto definitivo; podríamos haber dado más pasos. Se usa el valor absoluto $|b|$ para que la lógica funcione con divisores negativos.
* **Unicidad:** $q$ y $r$ son únicos. No existen dos formas distintas de dar la cantidad máxima de pasos $q$ de tamaño $b$.
* **¿Por qué $0 \le r$?** Sabemos que $qb \le a \iff a - qb \in \mathbb{Z}_+$. Y dado que $a - qb = r$, entonces obligatoriamente $r \in \mathbb{Z}_+$.

**Intuición en la Recta Numérica:** Partiendo de una posición inicial, quiero acercarme lo máximo posible a $a$ dando pasos de tamaño $b$ sin pasarme. El conjunto de pasos válidos es $A = \{k \cdot b \mid k \cdot b \le a, k \in \mathbb{N}^+\}$. El cociente $q$ nos da el salto máximo: $q \cdot b = \max(A)$, de manera que el siguiente salto ya se pasa: $(q+1) \cdot b > a$.

## Relación de Orden Parcial: "Divide a" ($|$)

Hasta ahora, hemos visto un solo tipo de relación de orden ($\le$) basado en la suma, dando lugar al conjunto totalmente ordenado $(\mathbb{Z}, \le)$, visualizable en una recta numérica.

La relación **"divide" ($|$)** es un orden fundamentado en el producto:
$$b \mid a \iff \exists q \in \mathbb{Z}_+ \mid a = q \cdot b$$

Esto da lugar al **conjunto parcialmente ordenado** $(\mathbb{Z}_+, \mid)$, que se visualiza de forma óptima mediante un retículo de divisibilidad o Diagrama de Hasse.

### Restricción a $\mathbb{Z}_+$
La relación $b \mid a$ funciona como un orden parcial estricto solo en $\mathbb{Z}_+$. Si permitiéramos todo el conjunto $\mathbb{Z}$, se incumpliría la **propiedad antisimétrica**.
Por ejemplo:
* $a \mid -a \iff -a = (-1) \cdot a$
* $-a \mid a \iff a = (-1) \cdot -a$
Por lo tanto, $a$ precede a $-a$ y, a la vez, lo sucede. Para mantener la jerarquía de orden, restringimos el análisis a $\mathbb{Z}_+$.

## Propiedades de la Relación de Divisibilidad

Para todo $a, b, c \in \mathbb{Z}_+$, el orden parcial $(\mathbb{Z}_+, \mid)$ cumple:

1.  **Reflexiva:** $a \mid a \iff a = 1 \cdot a$
2.  **Antisimétrica:** $b \mid a \land a \mid b \implies a = b$
    *Demostración:* $b \mid a \implies a = m \cdot b$,
    $a \mid b \implies b = n \cdot a$
    Sustituyendo: $b = n \cdot (m \cdot b)$. Por la propiedad cancelativa ($b \neq 0$), obtenemos $1 = n \cdot m$. En $\mathbb{Z}_+$, esto solo es posible si $n=1 \land m=1$, lo que implica $a = b$.
3.  **Transitiva:** $a \mid b \land b \mid c \implies a \mid c$
    *Demostración:*
    $b = q_1 \cdot a$,
    $c = q_2 \cdot b \implies c = q_2 \cdot (q_1 \cdot a) = (q_2 \cdot q_1) \cdot a \iff a \mid c$

### Cotas del Retículo: El Cero y el Uno
De estas propiedades derivan dos consecuencias estructurales importantes:
1.  **El $0$ es divisible por todos:** $a \mid 0 \iff 0 = 0 \cdot a$. En el orden parcial $(\mathbb{Z}_+, \mid)$, el cero es el elemento **maximal** y único máximo absoluto: $0 = \max((\mathbb{Z}_+, \mid))$.
2.  **El $1$ es divisor de todos:** $1 \mid a \iff a = 1 \cdot a$. El uno es el elemento **minimal** y único mínimo absoluto: $1 = \min((\mathbb{Z}_+, \mid))$.

## Suma de Ideales e Identidad de Bézout

La suma algebraica de dos ideales se define como el conjunto de todas las combinaciones lineales posibles de sus elementos:
$$a\mathbb{Z} + b\mathbb{Z} = \{a \cdot u + b \cdot v \mid u, v \in \mathbb{Z}\}$$

De esta definición extraemos que $a, b \in (a\mathbb{Z} + b\mathbb{Z})$:
* Si $u=0, v=1 \implies a\cdot 0 + b\cdot 1 = b$
* Si $u=1, v=0 \implies a\cdot 1 + b\cdot 0 = a$

Además, los propios ideales están contenidos enteros en la suma: $a\mathbb{Z}, b\mathbb{Z} \subset (a\mathbb{Z} + b\mathbb{Z})$.
Pero, ¿es $a\mathbb{Z} + b\mathbb{Z}$ un ideal principal generable por un solo número? La **Identidad de Bézout** lo responde a través del máximo común divisor.

### Máximo Común Divisor (m.c.d)
El máximo común divisor entre dos enteros es simplemente el **último ancestro común** en el árbol genealógico. Sea $d = \text{mcd}(a,b)$:
* $d \mid a \implies a = m \cdot d$
* $d \mid b \implies b = n \cdot d$

Si sustituimos esto en la suma de ideales:
$$a\mathbb{Z} + b\mathbb{Z} = \{m \cdot d \cdot u + n \cdot d \cdot v \mid u, v \in \mathbb{Z}\} = \{d \cdot (m\cdot u + n\cdot v) \mid u, v \in \mathbb{Z}\}$$

Como la suma $(m\cdot u + n\cdot v)$ es simplemente otro entero, queda demostrado que:
$$a\mathbb{Z} + b\mathbb{Z} = d\mathbb{Z}$$
**La suma de dos ideales principales resulta en otro ideal principal**, cuyo generador $d$ es el ancestro común (m.c.d).

## Conclusiones sobre la Identidad de Bézout

A partir del análisis de la suma de ideales, obtenemos la Identidad de Bézout:
$$\exists u, v \in \mathbb{Z} \mid \text{mcd}(a,b) = a \cdot u + b \cdot v$$

* Siempre existe una combinación lineal entre $a$ y $b$ que da como resultado su máximo común divisor.
* *(Aclaración técnica)*: Los coeficientes $u$ y $v$ no son únicos; hay infinitos pares que satisfacen la ecuación diofántica.
* Relacionado al orden parcial, el m.c.d es el mínimo del conjunto generado por la suma de los ideales:
    $$\text{mcd}(a,b) = \min((a\mathbb{Z} + b\mathbb{Z}, \mid)) = \min((d\mathbb{Z}, \mid)) = d$$
* Como la suma $a\mathbb{Z} + b\mathbb{Z} = d\mathbb{Z}$ siempre existe, podemos asegurar que todo par de números tiene un único ancestro común. Si son primos relativos (e.g., el $2$ y el $3$), su ancestro será el $1$: $2\mathbb{Z} + 3\mathbb{Z} = 1\mathbb{Z} = \mathbb{Z}$.

# El Algoritmo de Euclides y la Identidad de Bézout

Si volvemos a la definición de división entera ($a = q \cdot b + r$), es sencillo ver su profundo parecido con la **Identidad de Bézout**.

Sabemos que la suma de ideales es $a\mathbb{Z} + b\mathbb{Z} = \{a\cdot u + b\cdot v \mid u,v \in \mathbb{Z}\}$.
Si tomamos $u = 1$ y $v = -q$, obtenemos:
$$a \cdot (1) + b \cdot (-q) = a - q \cdot b = r$$
Esto demuestra que el resto $r$ pertenece a la suma de ideales: $r \in a\mathbb{Z} + b\mathbb{Z}$.

De forma recíproca, si reordenamos la división entera como $a = b \cdot q + r \cdot 1$, vemos que $a$ es una combinación lineal de $b$ y $r$, por lo que $a \in b\mathbb{Z} + r\mathbb{Z}$.

Si $a \in b\mathbb{Z} + r\mathbb{Z}$, toda la "descendencia" o hijos de $a$ están contenidos en ese ideal. Es decir, $a\mathbb{Z} \subset b\mathbb{Z} + r\mathbb{Z}$. Por la misma lógica aplicada al resto, $r\mathbb{Z} \subset a\mathbb{Z} + b\mathbb{Z}$.

De esta doble inclusión se hace evidente la siguiente igualdad fundamental:
$$a\mathbb{Z} + b\mathbb{Z} = b\mathbb{Z} + r\mathbb{Z} = d\mathbb{Z}$$
Donde $d$ es el máximo común divisor (m.c.d).

**Conclusión:** El algoritmo de Euclides parte del hecho algebraico de que:
$$\text{mcd}(a,b) = \text{mcd}(b,r) \quad \text{dado que } a = q \cdot b + r$$
En otras palabras, todo divisor de $b$ y $r$ es también divisor de $a$; todo ancestro común de $b$ y $r$ es ancestro común de $a$.

## Estructura Recursiva del Algoritmo

El algoritmo de Euclides es inherentemente recursivo. Su lógica paso a paso es:

1. **Caso base:** Si dos números enteros $a$ y $b$ son múltiplos ($b \mid a$), el ancestro común es simplemente el menor de los dos.
   * `if` $b \mid a \land a > b \implies \text{mcd}(a,b) = b$
2. **Caso recursivo:** Si no lo son ($b \nmid a$), existe un cociente $q$ y un resto $r$ tal que $a = q\cdot b + r$. Las parejas $(a,b)$ y $(b,r)$ tienen el mismo ancestro común.
   * `else` $b \nmid a \implies a = q\cdot b + r \quad (0 < r < b)$
   * $\implies \text{mcd}(a,b) = \text{mcd}(b,r)$

**¿Dónde está la recursividad?**
Cuando nos damos cuenta de que podemos aplicar la misma lógica al nuevo par $(b,r)$:
* `if` $r \mid b \land b > r \implies \text{mcd}(b,r) = r$
* `else` $r \nmid b \implies b = q' \cdot r + r' \quad (0 < r' < r < b)$
  * $\implies \text{mcd}(b,r) = \text{mcd}(r,r')$

Y seguimos iterando:
* `if` $r' \mid r \land r > r' \implies \text{mcd}(r,r') = r'$
* `else` $r' \nmid r \implies r = q'' \cdot r' + r'' \quad (0 < r'' < r' < r < b)$
  * $\implies \text{mcd}(r,r') = \text{mcd}(r',r'')$

Al hacer esto, estamos construyendo un conjunto de residuos $\{r, r', r'', \dots\}$ que se hace estrictamente más pequeño en cada iteración ($r > r' > r'' \dots$).
El algoritmo establece que el m.c.d es el mínimo de este conjunto de residuos no nulos:
$$\text{mcd}(a,b) = \min(\{r, r', r'', \dots\})$$
Es decir, **el m.c.d es el último residuo $r$ no nulo** obtenido en las sucesivas divisiones enteras.

$$a \cdot u + b \cdot v = 1$$
Esta expresión es fundamental y se conoce como **Ecuación Diofántica Lineal**.

Sin embargo, los coeficientes $u$ y $v$ en esta ecuación no son únicos. Existe un conjunto infinito de pares $\{(u,v), (u',v'), (u'',v'') \dots\}$ que permiten que $ax + by = 1$.

¿Cómo podemos hallar todas las soluciones posibles para $u$ y $v$?

## El Teorema de Gauss (Lema de Euclides)

Para hallar todas las soluciones enteras de una Ecuación Diofántica Lineal, utilizamos un principio conocido como el Teorema de Gauss. Este establece que:

$$\text{Si } \text{mcd}(a,b) = 1 \text{ (son coprimos) y } a \mid bc \implies a \mid c$$

### Intuición Genealógica: "Alguien se ha casado con su tío"
¿Por qué funciona esto? Imaginemos la relación de divisibilidad como lazos familiares:
Si $a$ y $b$ son hermanos (hijos directos del 1, es decir, coprimos), y a la vez resulta que $a$ es padre/ancestro de $bc$ (el cual es hijo de $b$ y $c$), entonces obligatoriamente $c$ tiene que ser hijo de $a$.

> **Mnemotecnia:** Respecto a $a$: *"Si el hijo $bc$ de mi hermano $b$ es mi nieto, entonces su esposa $c$ tiene que ser mi hija".*

## Solución General de la Ecuación Diofántica Lineal

¿Cómo sirve el Teorema de Gauss para hallar todos los pares $(u,v)$ de la ecuación $a \cdot u + b \cdot v = 1$?

1. **Hallar la solución particular:**
   Usando el algoritmo de Euclides hacia atrás (Algoritmo de Euclides Extendido), podemos encontrar la primera pareja $(u,v)$. Sabemos que el último residuo no nulo es el m.c.d, por lo que podemos despejarlo iterativamente hasta obtener:
   $$1 = a \cdot u + b \cdot v$$
   Al menos una pareja $(u,v)$ siempre se puede encontrar de esta forma.

2. **Igualar para hallar la solución general:**
   Supongamos que $(x,y)$ es cualquier otra pareja solución del conjunto infinito. Podemos igualar la pareja conocida con la genérica:
   $$a \cdot u + b \cdot v = a \cdot x + b \cdot y$$
   Reagrupamos términos de $a$ y $b$:
   $$a \cdot u - a \cdot x = b \cdot y - b \cdot v$$
   $$a(u - x) = b(y - v)$$

3. **Aplicar el Teorema de Gauss:**
   La igualdad $a(u - x) = b(y - v)$ nos dice dos cosas:
   * $a \mid b(y - v)$
   * $b \mid a(u - x)$
   
   Centrémonos en la primera: $a \mid b(y - v)$. Como sabemos por hipótesis que $\text{mcd}(a,b) = 1$, el Teorema de Gauss dicta que obligatoriamente $a$ debe dividir al otro factor.
   $$a \mid b(y - v) \land \text{mcd}(a,b) = 1 \implies a \mid (y - v)$$

4. **Despejar las variables:**
   Si $a \mid (y - v)$, significa que $(y - v)$ es un múltiplo de $a$. Podemos expresarlo como:
   $$y - v = k \cdot a \quad (\text{para algún } k \in \mathbb{Z})$$
   Por lo tanto, despejando $y$:
   **$$y = v + k \cdot a$$**

   Ahora sustituimos esto en nuestra igualdad factorizada $a(u - x) = b(y - v)$:
   $$a(u - x) = b(k \cdot a)$$
   Cancelamos $a$ en ambos lados:
   $$u - x = b \cdot k$$
   Despejando $x$:
   **$$x = u - b \cdot k$$**

**Conclusión:**
Dada una solución particular $(u,v)$ hallada mediante el algoritmo de Euclides, **todas las soluciones enteras** de la ecuación diofántica lineal $ax + by = 1$ están dadas por el conjunto de pares:
$$(x, y) = (u - b \cdot k, \; v + a \cdot k) \quad \forall k \in \mathbb{Z}$$
